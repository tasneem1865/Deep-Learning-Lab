```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, TimeDistributed
from tensorflow.keras.utils import to_categorical
```

```python
sentences = [
    "book flight from mumbai to delhi",
    "fly from pune to bangalore",
    "flight from delhi to mumbai",
    "book ticket from chennai to delhi",
    "travel from mumbai to pune",
    "fly from bangalore to chennai",
    "book flight from pune to delhi",
    "travel from delhi to bangalore"
]

labels = [
    ["O","O","O","B-FROM","O","B-TO"],
    ["O","O","O","B-FROM","O","B-TO"],
    ["O","O","O","B-FROM","O","B-TO"],
    ["O","O","O","B-FROM","O","B-TO"],
    ["O","O","O","B-FROM","O","B-TO"],
    ["O","O","O","B-FROM","O","B-TO"],
    ["O","O","O","B-FROM","O","B-TO"],
    ["O","O","O","B-FROM","O","B-TO"]
]

print("Number of sentences:", len(sentences))
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)

X = tokenizer.texts_to_sequences(sentences)

print("Word Index:")
print(tokenizer.word_index)
```

```python
label_to_id = {
    "O": 0,
    "B-FROM": 1,
    "B-TO": 2
}

y = [[label_to_id[label] for label in sequence] for sequence in labels]

print(y)
```

```python
max_len = max(len(seq) for seq in X)

X = pad_sequences(X, maxlen=max_len, padding='post')
y = pad_sequences(y, maxlen=max_len, padding='post')

print("Input shape:", X.shape)
print("Label shape:", y.shape)
```

```python
num_classes = 3

y = to_categorical(y, num_classes=num_classes)

print("Final label shape:", y.shape)
```

```python
vocab_size = len(tokenizer.word_index) + 1

model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=32, input_length=max_len),
    LSTM(64, return_sequences=True),
    TimeDistributed(Dense(num_classes, activation='softmax'))
])
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
history = model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)
```

```python
loss, accuracy = model.evaluate(X, y, verbose=0)

print("Accuracy:", accuracy * 100, "%")
print("Loss:", loss)
```

```python
test_sentence = "book flight from mumbai to delhi"

sequence = tokenizer.texts_to_sequences([test_sentence])
sequence = pad_sequences(sequence, maxlen=max_len, padding='post')

prediction = model.predict(sequence, verbose=0)
predicted_labels = np.argmax(prediction, axis=-1)[0]
```

```python
words = test_sentence.split()

id_to_label = {
    0: "O",
    1: "B-FROM",
    2: "B-TO"
}

for word, label_id in zip(words, predicted_labels):
    print(word, "->", id_to_label[label_id])
```

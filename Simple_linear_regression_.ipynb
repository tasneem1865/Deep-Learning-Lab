```python
# Tasneem Sadikot
# 23070521162

import numpy as np
import matplotlib.pyplot as plt

x = np.array([100, 150, 200, 250, 300])
y = np.array([20, 25, 30, 35, 40])

print("Input X:", x)
print("Input Y:", y)
```

```python
mean_x = np.mean(x)
mean_y = np.mean(y)

print("Mean of X:", mean_x)
print("Mean of Y:", mean_y)
```

```python
numerator = np.sum((x - mean_x) * (y - mean_y))
denominator = np.sum((x - mean_x)**2)

b1 = numerator / denominator

print("Slope (b1):", b1)
```

```python
b0 = mean_y - b1 * mean_x

print("Intercept (b0):", b0)
```

```python
y_pred = b0 + b1 * x

print("Original Y values:", y)
print("Predicted Y values (y_pred):", np.round(y_pred, 2))
```

```python
residuals = y - y_pred

print("Residuals (Actual Y - Predicted Y):", np.round(residuals, 2))
```

```python
mae = np.mean(np.abs(residuals))
mse = np.mean(residuals**2)
rmse = np.sqrt(mse)

print(f"Manual MAE: {mae:.2f}")
print(f"Manual MSE: {mse:.2f}")
print(f"Manual RMSE: {rmse:.2f}")
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='blue', label='Actual Data Points')
plt.plot(x, y_pred, color='red', label='Regression Line')
plt.title('Simple Linear Regression')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

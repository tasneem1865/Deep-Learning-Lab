```python
# Install mlxtend if not already installed
!pip install mlxtend -q

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Create a new transactional dataset with requested items
data = {
    'Transaction_ID': [1, 1, 1, 2, 2, 2, 3, 3, 4, 4, 4, 4, 5, 5],
    'Item': [
        'Milk', 'Bread', 'Butter',
        'Bread', 'Beer', 'Diaper',
        'Milk', 'Diaper',
        'Milk', 'Bread', 'Beer', 'Diaper',
        'Bread', 'Butter'
    ]
}

df_raw = pd.DataFrame(data)
print("--- Raw Transactions ---")
display(df_raw.head(10))

# Pivot transactions into a binary matrix using boolean values
basket = (df_raw.groupby(['Transaction_ID', 'Item'])['Item']
          .count().unstack().reset_index().fillna(0)
          .set_index('Transaction_ID'))

basket = basket.map(lambda x: True if x > 0 else False)
print("\n--- One-Hot Encoded Basket ---")
display(basket)

# Generate frequent itemsets
frequent_itemsets = apriori(basket, min_support=0.3, use_colnames=True)
print("\n--- Frequent Itemsets ---")
display(frequent_itemsets)

# Generate association rules
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)
print("\n--- Association Rules ---")
display(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Create a sample transactional dataset
data = {
    'Transaction_ID': [1, 1, 1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 5, 5],
    'Item': ['Milk', 'Bread', 'Butter', 'Bread', 'Butter', 'Milk', 'Bread', 'Butter', 'Milk', 'Bread', 'Milk', 'Bread', 'Butter', 'Cookies']
}

df_raw = pd.DataFrame(data)
display(df_raw.head(10))
```

```python
# Pivot transactions into a binary matrix
basket = (df_raw.groupby(['Transaction_ID', 'Item'])['Item']
          .count().unstack().reset_index().fillna(0)
          .set_index('Transaction_ID'))

# Convert values to boolean/binary (True/False or 1/0)
basket = basket.map(lambda x: 1 if x > 0 else 0)
display(basket)
```

```python
# Generate frequent itemsets with support >= 0.4 (40% of transactions)
frequent_itemsets = apriori(basket, min_support=0.4, use_colnames=True)
print("--- Frequent Itemsets ---")
display(frequent_itemsets)

# Generate association rules with a confidence threshold >= 0.7
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)
print("\n--- Association Rules ---")
display(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```
